```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm=ChatGoogleGenerativeAI(model='gemini-2.5-flash')

prompt='Tell me somethig about Lion'

response=llm.invoke(prompt)
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
llm=ChatGoogleGenerativeAI(model='gemini-2.5-flash')
prompt=PromptTemplate.from_template('What is {Topic}')

chain=prompt | llm

response=chain.invoke({"Topic":"UPSC?"})
print(response)
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(model='gemini-2.5-flash',temperature=0.2)

prompt="Persian Cat's Origin"
response=llm.invoke(prompt)

print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate

prompt=PromptTemplate.from_template("Define {Topic}")

chain=prompt | llm

Topic=input("Enter Topic to Find Information of ? ")

response=chain.invoke({"Topic":Topic})
```

```python
print(response.content)
```
